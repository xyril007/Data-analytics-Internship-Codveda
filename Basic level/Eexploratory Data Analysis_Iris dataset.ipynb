```python
#Checking  for missing Values#
```

```python
plt.Figure(figsize=(10,6))
sns.heatmap(df.isnull(), cbar =False ,cmap='viridis')
plt.title("Missing Values")
plt.show()
```

```python
#Checking Duplicates values and generating figures for duplicates values
```

```python
duplicates = df[df.duplicated(keep=False)]
duplicates
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.duplicated().to_frame(), cbar=False, cmap='viridis')
plt.title("Duplicate Rows")
plt.show()
```

```python
#Rearranging clumns
```

```python
df = df[[df.columns[-1]] + list(df.columns[:-1])]
```

```python
df.columns
```

```python
plt.figure(figsize=(12,5))
sns.histplot(df['sepal_length'].dropna(),kde=True, bins=30,color= "orange")
plt.title("Sepal_length Distribution")
plt.xlabel("sepal_length")
plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.histplot(df['sepal_width'].dropna(), kde=True, bins=30, color="blue")

plt.title("Sepal Width Distribution")
plt.xlabel("Sepal Width")
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.histplot(df['petal_length'].dropna(), kde=True, bins=30, color="blue")
plt.title("petal_length Distribution")
plt.xlabel("petal_length")
plt.show()
```

```python
plt.figure(figsize=(12, 5))
sns.histplot(df['petal_width'].dropna(), kde=True, bins=30, color="blue")
plt.title("petal_width Distribution")
plt.xlabel("petal_width")
plt.show()
```

```python
#Categorical value analysis
```

```python
df.columns
```

```python
plt.figure(figsize=(15,5))
sns.countplot(
    x='species', 
    data=df, 
    hue='species', 
    palette= 'Set2', 
    legend=False
)

plt.title("species_count")
plt.xlabel("species")
plt.ylabel("count")
plt.show()
```

```python
#Bivariate analysis
```

```python
corr_matrix=df.corr(numeric_only=True)
plt.figure(figsize=(10, 5))
sns.heatmap(corr_matrix), __annotations__=True, cmap= 'blue'
plt.title("correlation between species length and width")
plt.show()
```

```python
corr_matrix=df.corr(numeric_only=True)
plt.figure(figsize=(10, 5))
sns.heatmap(corr_matrix, annot=False, cmap= 'coolwarm')
plt.title("correlation between species length and width")
plt.show()
```
